```python
import ast
def convert(obj):
  L=list()
  for i in ast.literal_eval(obj):
    L.append(i['name'])
  return L
```

```python
import ast
def convertCast(obj):
  L=list()
  counter=0
  for i in ast.literal_eval(obj):
    if counter!=3:
      L.append(i['name'])
      counter=counter+1
    else:
      break;
  return L

movies_db['cast']=movies_db['cast'].apply(convertCast)
movies_db['cast']
```

```python
def fetch_director(obj):
  L=list()
  for i in ast.literal_eval(obj):
    if i['job']=='Director':
      L.append(i['name'])
      break;
  return L

movies_db['crew']=movies_db['crew'].apply(fetch_director)
movies_db['crew']
```

To implement Stemming importing nltk

```python
import nltk
```

```python
from nltk.stem.porter import PorterStemmer
ps=PorterStemmer()
```

```python
def Stem(text):
  y=list()

  for i in text.split():
    y.append(ps.stem(i))

  return " ".join(y)
```

```python
new_df['tags']=new_df['tags'].apply(Stem)
```

```python
from sklearn.feature_extraction.text import CountVectorizer
cv=CountVectorizer(max_features=5000,stop_words='english')
vectors=cv.fit_transform(new_df['tags']).toarray()
vectors
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity = cosine_similarity(vectors)
```

```python
print(similarity)
```

```python
similarity.shape
```

```python
similarity[0]
```

```python
def recommend(movie, top_n=5):
  movie_index=new_df[new_df['title']==movie].index[0]
  distances=similarity[movie_index]
  movies_list=sorted(list(enumerate(distances)),reverse=True, key=lambda x:x[1])

  recommended_movies_list = []
  for i in movies_list[1:top_n+1]:
    recommended_movies_list.append(new_df.iloc[i[0]].title)
  return recommended_movies_list
```

```python
recommend('Batman Begins')
```

```python
movie_genres = {}

for _, row in movies.iterrows():
    movie_genres[row['title']] = set(row['genres'])
```

```python
def genre_overlap_score(query_movie, recommended_movies):
    query_genres = movie_genres[query_movie]

    scores = []
    for movie in recommended_movies:
        overlap = query_genres.intersection(movie_genres[movie])
        score = len(overlap) / len(query_genres) if query_genres else 0
        scores.append(score)

    return scores
```

```python
def precision_at_k(query_movie, recommended_movies, k=5):
    query_genres = movie_genres[query_movie]
    hits = 0

    for movie in recommended_movies[:k]:
        if query_genres.intersection(movie_genres[movie]):
            hits += 1

    return hits / k
```

```python
def hit_rate(query_movie, recommended_movies):
    query_genres = movie_genres[query_movie]

    for movie in recommended_movies:
        if query_genres.intersection(movie_genres[movie]):
            return 1
    return 0
```

```python
query_movie = "Avatar"   # change this if needed

recommended_movies = recommend(query_movie, top_n=5)

print("Query Movie:", query_movie)
print("Recommended Movies:", recommended_movies)
print("Precision@5:", precision_at_k(query_movie, recommended_movies, k=5))
print("Genre Overlap Scores:", genre_overlap_score(query_movie, recommended_movies))
print("Hit Rate:", hit_rate(query_movie, recommended_movies))
```
